# Text-to-SQL Analytics Exploration
Explore the sample developer-jobs dataset with SQLite and pandas before using the Streamlit dashboard.

```python
import sqlite3
import pandas as pd
from pathlib import Path

data = pd.DataFrame([
    ["France", "Data Scientist", "Python", 62000, 180],
    ["Germany", "Data Scientist", "Python", 72000, 240],
    ["Italy", "Data Analyst", "SQL", 42000, 190],
    ["Spain", "Software Engineer", "Python", 43000, 205],
    ["Netherlands", "ML Engineer", "PyTorch", 78000, 130],
], columns=["country", "role", "technology", "average_salary_eur", "job_count"])

conn = sqlite3.connect(":memory:")
data.to_sql("developer_jobs", conn, index=False, if_exists="replace")
```

```python
query = """
SELECT country,
       ROUND(AVG(average_salary_eur), 0) AS average_salary_eur,
       SUM(job_count) AS total_jobs
FROM developer_jobs
GROUP BY country
ORDER BY average_salary_eur DESC
"""
result = pd.read_sql_query(query, conn)
result
```

```python
import plotly.express as px

fig = px.bar(
    result,
    x="country",
    y="average_salary_eur",
    title="Average Salary by Country",
)
fig.show()
```
